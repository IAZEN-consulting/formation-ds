## Mise en situation

La reconnaissance optique de caractères (OCR) consiste à identifier automatiquement des
caractères à partir d'images. C'est un problème classique pour s'entraîner à formuler un
problème de Data Science.

Ce TP est volontairement un exercice de **formulation**, pas de code : à ce stade de la
journée, vous n'avez pas encore vu Python ni Scikit-learn. L'objectif est de traduire un
problème métier en problème de machine learning, avec le bon vocabulaire.

```python
from sklearn.datasets import load_digits

import matplotlib.pyplot as plt
digits = load_digits()

fig, axes = plt.subplots(1, 10, figsize=(8, 2))
for ax, image, label in zip(axes, digits.images, digits.target):
    ax.imshow(image, cmap="gray_r")
    ax.set_title(label)
    ax.axis("off")
plt.show()
```

## Étape 1 — Formuler la question métier

**Question de réflexion :** en une phrase, quelle est la question à laquelle notre modèle
doit répondre ?


**Réponse commentée**

> « Quel caractère est représenté sur cette image ? »

C'est la reformulation, en langage métier, du besoin : on part toujours d'une question
compréhensible par un non-spécialiste avant de la traduire en objet mathématique.


## Étape 2 — Identifier la nature du problème

**Question de réflexion :** s'agit-il de classification ou de régression ? Binaire ou
multi-classes ?


**Réponse commentée**

C'est un problème de **classification multi-classes** : il y a une classe possible par
caractère (par exemple les 10 chiffres 0 à 9, ou les 26 lettres A à Z), et la target est
catégorielle, pas continue.


## Étape 3 — Définir l'input

**Question de réflexion :** une image n'est pas un tableau de nombres au sens où
Scikit-learn l'attend. Comment la transformer en features exploitables par un modèle ?


**Réponse commentée**

Chaque image est transformée en un **vecteur de pixels** : on aplatit la grille de pixels
(par exemple une image 28x28 devient un vecteur de 784 valeurs), et chaque pixel devient
une feature. La matrice X a donc une ligne par image et une colonne par pixel.


## Étape 4 — Définir l'output

**Question de réflexion :** que doit renvoyer le modèle pour une nouvelle image ?


**Réponse commentée**

La classe prédite parmi l'ensemble des caractères possibles (par exemple un chiffre entre
0 et 9), éventuellement accompagnée d'une probabilité par classe.


## Étape 5 — Discussion en groupe

**Question de réflexion :** quelles difficultés pratiques peut-on rencontrer, et quelles
features seraient utiles pour les limiter ?

Pistes de discussion à animer avec le groupe (pas de réponse unique) :

- Bruit dans l'image, qualité de scan
- Rotation ou inclinaison du caractère
- Variabilité des polices ou de l'écriture manuscrite
- Caractères qui se ressemblent (le chiffre 1 et la lettre I, le 0 et le O)
- Features possibles : normalisation de la taille et du contraste, détection de contours,
